```python
import pandas as pd
import numpy as np
from google.colab import files

uploaded = files.upload()
```

```python
import io
# Charger le CSV dans un DataFrame
filename = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[filename]))

# Afficher les premières lignes
display(df.head(15))
```

```python
# Nombre de répondants
print("Nombre de répondants :", df["Respondent ID"].nunique())

# Vérifier les doublons
print("Doublons :", df.duplicated().sum())

# Vérifier les réponses aux situations de choix
choice_cols = [col for col in df.columns if "Situation" in col]

display(df[choice_cols].head())
```

```python
col_age = "Quel âge avez-vous ?"

df = df[df[col_age] != "35+"].copy()

print("Dimensions après suppression :", df.shape)
```

```python
print(df[col_age].value_counts())
```

```python
# Dimensions du dataset
print("Dimensions :", df.shape)

# Types des variables
df.info()

# Nombre de valeurs manquantes par colonne
df.isna().sum()
```

```python
# 1. Supprimer la colonne Submission ID
df = df.drop(columns=["Submission ID"])

# 2. Recoder l'âge
df["Quel âge avez-vous ?"] = df["Quel âge avez-vous ?"].replace({
    "18-24 ans": 1,
    "25-34 ans": 2
})

# 3. Recoder le genre
df["À quel genre vous identifiez-vous ?"] = df["À quel genre vous identifiez-vous ?"].replace({
    "Homme": 1,
    "Femme": 2
})

# 4. Situation actuelle
df["Actuellement, vous êtes... ?"] = df["Actuellement, vous êtes... ?"].replace({
    "Etudiant": 1,

    "Jeune actif": 2
})

# Vérification
display(df.head())

print(df["Quel âge avez-vous ?"].value_counts(dropna=False))
print(df["À quel genre vous identifiez-vous ?"].value_counts(dropna=False))
```

```python
# 5. Région de résidence
df["Où résidez-vous principalement ?"] = df["Où résidez-vous principalement ?"].replace({
    "Suisse romande": 1,
    "Suisse alémanique": 2,
    "Suisse italienne": 3,
    "A l'étranger": 4
})

# 6. Achat personnel de vêtements
df["Achetez-vous vous-même vos vêtements ?"] = df["Achetez-vous vous-même vos vêtements ?"].replace({
    "Oui": 1,
    "Non": 2
})

# 7. Fréquence d'achat
df["À quelle fréquence achetez-vous de nouveaux vêtements (neufs ou d'occasion) ?"] = df[
    "À quelle fréquence achetez-vous de nouveaux vêtements (neufs ou d'occasion) ?"
].replace({
    "Toutes les semaines": 1,
    "Tous les mois": 2,
    "Quelques fois par an": 3
})

# 8. Budget disponible mensuel
df["De combien d'argent disposez-vous chaque mois après vos dépenses fixes (loyer, assurances, transports) ?"] = df[
    "De combien d'argent disposez-vous chaque mois après vos dépenses fixes (loyer, assurances, transports) ?"
].replace({
    "moins de 200.-": 1,
    "200 à 499 .-": 2,
    "500 à 1000 .-": 3,
    "plus de 1000.-": 4
})

# 9. Dépenses mensuelles en vêtements
df["Combien dépenser vous en moyenne par mois pour vos vêtements ?"] = df[
    "Combien dépenser vous en moyenne par mois pour vos vêtements ?"
].replace({
    "moins de 50.-": 1,
    "50 à 99 .-": 2,
    "100 à 200.-": 3,
    "plus de 200.-": 4
})
```

```python
df["Avez-vous acheté des vêtements de seconde main au cours des 12 derniers mois ?"] = df[
    "Avez-vous acheté des vêtements de seconde main au cours des 12 derniers mois ?"
].replace({
    "Oui": 1,
    "Non": 2
})

df["Où avez-vous fait votre achat ?"] = df[
    "Où avez-vous fait votre achat ?"
].replace({
    "Plateforme en ligne": 1,
    "Magasin": 2
})
```
